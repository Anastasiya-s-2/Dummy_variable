```python
import numpy as np
import statsmodels.api as sm

x_1 = np.array([16988.84, 13632.10, 12345.48, 9232.78, 6263.36, 1209.92, 6456.91, 4310.09, 6962.56, 5351.18])
y_1 = np.array([426.24, 342.32, 310.16, 232.33, 158.10, 31.76, 162.94, 109.27, 175.58, 135.30])

x_2 = np.array([131.91, 354.32, 632.04, 431.60, 67.97, 122.71, 225.29, 308.78, 189.06, 339.71])
y_2 = np.array([27.83, 72.36, 127.85, 87.73, 15.00, 25.96, 46.52, 63.23, 39.23, 69.39])

x_all = np.concatenate([x_1, x_2])
y_all = np.concatenate([y_1, y_2])

d = np.array([1] * 10 + [0] * 10)

desing_mat = np.column_stack([np.ones_like(x_all),x_all,d,x_all * d])

model = sm.OLS(y_all, desing_mat).fit()

print("Результати:")
print(model.summary())

print("\nКоефіцієнти:")
params = model.params
print(f"a0 (базовий вільний член --- точка перетину прямої з віссю ординат) = {params[0]:.4f}")
print(f"a1 тангенс кута нахилу до осі абсцис = {params[1]:.4f}")
print(f"beta_1 = {params[2]:.4f}")
print(f"beta_2 = {params[3]:.4f}")
print(f"Коефіцієнт детермінації R^2 = {model.rsquared:.4f}")
print("\nоцінка моделі: фіктивна зміна зрушення + фіктивна зміна нахилу")
print(f"y = {params[0]:.4f} + {params[1]:.4f}*x + {params[4] if len(params)>4 else params[2]:.4f}*d + {params[3]:.4f}*(x*d)")
print("Рівняння для середніх банків (d = 0)")
print(f"y = {params[0]:.3f} + {params[1]:.3f} * x")

print("Рівняння для великих банків (d = 1)")
intercept_large = params[0] + params[2]
slope_large = params[1] + params[3]
print(f"y = ({params[0]:.3f} + {params[2]:.3f}) + ({params[1]:.3f} + {params[3]:.3f}) * x")
print(f"y = {intercept_large:.3f} + {slope_large:.3f} * x")
```
